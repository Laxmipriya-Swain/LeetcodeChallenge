```python
#Print Name N times using Recursion
def print_name(name,n):
    if n == 0:
        return
    print(name)
    print_name(name,n-1)
# driver code
name = input("Enter name: ")
n = int(input("Enter N: "))
print_name(name, n)
```

```python
#Print 1 to N using Recursion (forward recursion)
def func(i, N):
    if i > N:
        return
    print(i)   # print FIRST
    func(i + 1, N)  # then recurse

N = int(input("Enter N: "))
func(1, N)   # start counting from 1
```

```python
#Print 1 to N using Recursion (Backtracking)
def func(i, N):
    if i > N:
        return
    func(i + 1, N)  # then recurse
    print(i)   # print FIRST

N = int(input("Enter N: "))
func(1, N)   # start counting from 1
```

```python
# Print N to 1 using Recursion (forward/direct printing)
def func(i):
    if i < 1:
        return
    print(i)          # print FIRST
    func(i - 1)        # then recurse, decreasing

N = int(input("Enter N: "))
func(N)   # start counting from N, going down
```

```python
# Print N to 1 using Recursion (forward/direct printing)
def func(i):
    if i < 1:
        return
    func(i - 1)        # then recurse, decreasing
    print(i)          # print FIRST

N = int(input("Enter N: "))
func(N)   # start counting from N, going down
```

```python
# Print N to 1 using Recursion (Backtracking)
def func(i, N):
    if i > N:
        return
    func(i + 1, N)   # recurse FIRST (go all the way down)
    print(i)          # print during backtracking (on the way back up)

N = int(input("Enter N: "))
func(1, N)
```

```python
#sum of 1 to N using parameterized Recursion
def func(sum, i, N):
    if i > N:
        print(sum)
        return
    func(sum + i, i + 1, N)

func(0, 1, 4)
```

```python
#sum of 1 to N using functional Recursion
def fun(N):
    if N == 1:
        return 1
    return N * fun(N-1)
N = int(input())
print(fun(N))
```

```python
#Factorial of a Number : Iterative Recursive
def factorial(num):
    if num == 0 or num == 1:
        return 1
    return num * factorial(num - 1)
num=int(input())
factorial(num)
```

```python
#Reverse of a array
def reverse_array(nums, left, right):
    if left >= right:
        return nums
    nums[left], nums[right] = nums[right], nums[left]
    return reverse_array(nums, left + 1, right - 1)

nums = [2, 4, 1, 7, 6, 3, 8, 9, 5]
print(reverse_array(nums, 0, len(nums) - 1))
```

```python
#Check if the given String is Palindrome or not
def func(s, left, right):
    if left >= right:
        return True
    if s[left] != s[right]:
        return False
    return func(s, left + 1, right - 1)

s = "abcba"
print(func(s, 0, len(s) - 1))   
```

```python
#Print Fibonacci Series up to Nth term
def fun(n):
    if n==0 or n==1:
        return 1
    return fun(n-1) + fun(n-2)
n=5
print(fun(n))
```

```python
#Print Fibonacci Series up to Nth term
def fibonacci(N):
    # Base case: if N is 0 or 1, return N
    if N <= 1:
        return N

    # Recursive calls: calculate previous two terms
    last = fibonacci(N - 1)   # (N-1)th term
    slast = fibonacci(N - 2)  # (N-2)th term

    return last + slast

# Driver code
N = 4
print(fibonacci(N))  
```
