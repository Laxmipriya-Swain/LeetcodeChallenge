```python
# Inverted Star Pyramid
m=int(input("enter a row number:"))
# Outer loop for rows
for i in range(m):
    # Print leading spaces
    for j in range(i):
        print(" ", end="")
    # Print stars
    for j in range(2*m - 2*i- 1):
        print("*",end="")
    # Print trailing spaces
    for j in range(i):
        print(" ",end="")
    # Move to next row
    print()
```

```python
#11.Binary Number Triangle Pattern
def pattern11(n):
    for i in range(n):
        if i%2 == 0:
            start=1
        else:
            start=0
        for j in range(i+1):
            print(start,end="")
            start=1-start
        print()

p = pattern11(n=5)
```

```python
#12.Number Crown Pattern
def pattern12(n):
    for i in range(1,n+1):
        for j in range(1,i+1):
            print(j,end=" ")
        for j in range(2*(n-i)):
            print(" ",end=" ")
        for j in range(i,0,-1):
            print(j,end=" ")
        print()
q=pattern12(5)
```

```python
# 13: Increasing Number Triangle Pattern
def pattern13(n):
    num = 1
    for i in range(1,n+1):
        for j in range(1,i+1):
            print(num,end=" ")
            num = num +1
        print()
L=pattern13(6)
```

```python

#16.Alpha-Ramp Pattern
def pattern15(n):
    for i in range(0,n+1):
        ch=chr(65+i)
        for j in range(0,i+1):
            print(ch,end=" ")
        print()
M = pattern15(5)
```

```python
#17.Alpha-Hill Pattern
def pattern17(N):

    for i in range(N):

        # Print spaces
        for j in range(N - i - 1):
            print(" ", end=" ")

        # Print increasing letters: A B C ...
        for j in range(i + 1):
            print(chr(65 + j), end=" ")

        # Print decreasing letters: ... C B A
        for j in range(i - 1, -1, -1):
            print(chr(65 + j), end=" ")

        print()
N = pattern17(5)
```

```python
#22.Symmetric-Void Pattern
def pattern22(n):
    #upper half
    for i in range(n):
      stars = n - i
      spaces = 2*i
      print("* " * stars + "  " * spaces + "* " * stars)
     #lower half
    for i in range(1,n+1):
       starts = i
       spaces = 2*(n-i)
       print("* " * starts + "  " * spaces + "* " * starts)
   
   
      
q=pattern22(5)
```

```python
#20.Butterfly
n = 5

# Upper half
for i in range(1, n + 1):
    # Left stars
    for j in range(i):
        print("*", end=" ")

    # Middle spaces
    for j in range(2 * (n - i)):
        print(" ", end=" ")

    # Right stars
    for j in range(i):
        print("*", end=" ")

    print()

# Lower half
for i in range(n - 1, 0, -1):
    # Left stars
    for j in range(i):
        print("*", end=" ")

    # Middle spaces
    for j in range(2 * (n - i)):
        print(" ", end=" ")

    # Right stars
    for j in range(i):
        print("*", end=" ")

    print()
```
