```python
#store frequency in dictionary
n = int(input())
arr = list(map(int,input().split()))
freq = {}
for i in arr:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1
q = int(input())
for j in range(q):
    num = int(input())
    print(freq.get(num, 0))
```

```python
n = int(input())
arr = list(map(int, input().split()))

hash_list = [0] * 11

for num in arr:
    hash_list[num] += 1

m = int(input())

for _ in range(m):
    num = int(input())

    if num < 1 or num > 10:
        print(0)
    else:
        print(hash_list[num])
```

```python
#charecter frequency
s = input()

# precompute
hash_list = [0] * 26

for ch in s:
    ascii_val = ord(ch)
    index = ascii_val - 97
    hash_list[index] += 1

q = int(input())

for _ in range(q):
    ch = input()
    ascii_val = ord(ch)
    index = ascii_val - 97
    print(hash_list[index])
```

```python
n = [5, 3, 2, 2, 1, 5, 5, 1, 5, 10]
m = [10, 11, 1, 9, 5, 6, 2]

# Precompute
hash_list = [0] * 11

for num in n:
    hash_list[num] += 1

# Queries
for num in m:
    if num < 1 or num > 10:
        print(0)
    else:
        print(hash_list[num])
```

```python
s = "aabczyza"
q = ["a", "z", "y", "b", "x"]

# Precompute
hash_list = [0] * 26

for ch in s:
    ascii_val = ord(ch)
    index = ascii_val - 97
    hash_list[index] += 1

# Queries
for ch in q:
    ascii_val = ord(ch)
    index = ascii_val - 97

    print(hash_list[index])
```

```python
s = "abcdabehf"
q = ["a", "g", "h", "b", "c"]

# Precompute
hash_list = [0] * 256

for ch in s:
    ascii_val = ord(ch)
    hash_list[ascii_val] += 1

# Queries
for ch in q:
    ascii_val = ord(ch)

    print(hash_list[ascii_val])
```
