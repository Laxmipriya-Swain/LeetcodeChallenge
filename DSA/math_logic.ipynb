```python
#Leetcode no -7
#Reverse of a number
class Solution(object):
    def reverse(self, x):
        n=abs(x)
        revnum=0
        while n > 0:
            lastdigit = n%10
            revnum=revnum *10 + lastdigit
            n = n//10
        #for -ve checking
        if x<0:
            revnum = -revnum
        if -2**31 <= revnum <= 2**31-1:
            return revnum
        return 0

sol = Solution()        # create an instance
j = sol.reverse(568)    # call the method on the instance
print(j)                 # 865
```

```python
#leet code - 9
#Palindrome
class Solution:
    def isPalindrome(self, x):
        num = x
        rev = 0
        if num < 0:
            return False
        while num > 0:
            ld = num % 10
            rev = rev * 10 + ld
            num = num // 10
        return x == rev
lol = Solution()
L =lol.isPalindrome(121)
L
```

```python
#print all divisors of a given number
from math import sqrt
n = int(input(" Enter a number: ")) 
res = []
for i in range(1, int(sqrt(n)) + 1):
    if n % i == 0:
        res.append(i)
        if i != n // i:
            res.append(n // i)
res.sort()
print("Divisors of", n, "are:", res)
```

```python
# Check if a number is Armstrong Number or not

def armstrongNumber (n):
    num = n
    nod = len(str(n))
    total = 0
    while num > 0:
        LD = num % 10
        total = total + LD ** nod
        num = num //10
    return n == total
n = 153
j=armstrongNumber(n)
j
```

```python
#prime
n = int(input("Enter a number:"))
if n > 1:
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            print(n, "is not a prime number")
            break
    else:
        print(n, "is a prime number")
```

```python
#GCD/HCF
a = int(input('enter a number: '))
b = int(input('enter a number: '))

while a > 0 and b > 0:
    if a > b:
        a = a % b
    else:
        b = b % a

print("GCD is:", a if a > 0 else b)
```
